# Lung Mask Generation — Full Dataset Pipeline (v3)

## What this notebook does
- Reads every `images/*.nii.gz` from **`hourouu/LIDC_IDRI_PROCESSED`** (source, read-only)
- Writes masks and ledger to **`hourouu/LIDC_IDRI_MASKS`** (destination, create it manually before first run)
- Processes files in batches of 10 — one commit per batch
- On every startup, pulls the ledger from the destination Hub repo — survives full session wipes

```python
# ── 1. Imports ─────────────────────────────────────────────────────────────────
import gc
import io
import json
import shutil
import time
import warnings
from pathlib import Path

import nibabel as nib
import numpy as np
from huggingface_hub import CommitOperationAdd, HfApi, HfFileSystem, hf_hub_download
from huggingface_hub.utils import EntryNotFoundError, RepositoryNotFoundError
from scipy import ndimage as ndi
from skimage import measure, morphology
from skimage.segmentation import clear_border
```

```python
# ── 2. Configuration ───────────────────────────────────────────────────────────
import os
try:
    from google.colab import userdata
    HF_TOKEN = userdata.get("HF_TOKEN")
except Exception:
    HF_TOKEN = os.environ.get("HF_TOKEN", "")  # fallback for non-Colab environments

REPO_TYPE  = "dataset"

# Source repo — where the raw images live (read-only)
SRC_REPO_ID = "hourouu/LIDC_IDRI_PROCESSED"
IMAGES_DIR  = "images"

# Destination repo — where masks and the ledger are written
# Created automatically on first run if it doesn't exist yet.
DST_REPO_ID      = "hourouu/LIDC_IDRI_MASKS"
MASKS_DIR        = "masks"
LEDGER_REPO_PATH = "masks/progress.json"   # lives ON THE HUB — survives session wipes

BATCH_SIZE = 10

# Local scratch — treated as fully ephemeral, nothing here is trusted across sessions
SCRATCH_DIR = Path("/content/scratch")
MASKS_OUT   = SCRATCH_DIR / "masks"
HF_CACHE    = SCRATCH_DIR / "hf"
for d in (SCRATCH_DIR, MASKS_OUT, HF_CACHE):
    d.mkdir(parents=True, exist_ok=True)

# Retry settings
MAX_RETRIES   = 3
RETRY_BACKOFF = [5, 15, 45]   # seconds between attempts

# Segmentation hyper-parameters
HU_THRESHOLD          = -400
DISK_CLOSING_RADIUS   = 3
DISK_CLOSING_TOUCHING = 5
SEAM_HEAL_RADIUS      = 2
MIN_BLOB_VOXELS       = 50
MIN_LUNG_FRACTION     = 0.15
MAX_LUNG_REGIONS      = 3
CLOSING_RADIUS        = 3
MIN_MASK_VOXELS       = 1000   # below this the mask is considered empty / bad

print(f"Source repo : {SRC_REPO_ID}  ({IMAGES_DIR}/)")
print(f"Dest repo   : {DST_REPO_ID}  ({MASKS_DIR}/)  ← will be auto-created if missing")
print(f"Ledger      : {DST_REPO_ID}/{LEDGER_REPO_PATH}")
print(f"Batch size  : {BATCH_SIZE}")
print(f"Scratch     : {SCRATCH_DIR}  (ephemeral)")
```

```python
# ── 3. Hub client + retry wrapper ─────────────────────────────────────────────

api = HfApi(token=HF_TOKEN)
fs  = HfFileSystem(token=HF_TOKEN)


def with_retry(fn, *args, label="operation", no_retry=(), **kwargs):
    """
    Call fn(*args, **kwargs) up to MAX_RETRIES times.
    Exceptions listed in `no_retry` are re-raised immediately without retrying
    (e.g. EntryNotFoundError for a missing ledger on first run).
    All other exceptions are retried with exponential backoff.
    Raises the last exception if all attempts fail.
    """
    last_exc = None
    for attempt in range(MAX_RETRIES):
        try:
            return fn(*args, **kwargs)
        except no_retry:
            raise   # not a transient error — don't retry
        except Exception as exc:
            last_exc = exc
            wait = RETRY_BACKOFF[min(attempt, len(RETRY_BACKOFF) - 1)]
            print(f"    {label} failed (attempt {attempt + 1}/{MAX_RETRIES}): {exc}")
            if attempt < MAX_RETRIES - 1:
                print(f"     Retrying in {wait}s…")
                import time; time.sleep(wait)
    raise last_exc
```

```python
# ── 4. Ledger — Hub-backed, survives session wipes ────────────────────────────
#
# The ledger is a JSON file stored at LEDGER_REPO_PATH inside the HF dataset.
# On every startup we pull it from the Hub. On every successful batch commit
# we push an updated copy alongside the masks in the same commit.
# The local copy is only a working buffer — never the source of truth.

def _empty_ledger() -> dict:
    return {"uploaded": [], "failed": [], "last_batch": -1}


def pull_ledger() -> dict:
    """
    Download the ledger from the Hub.
    Falls back to an empty ledger if the file doesn't exist yet.
    Raises on genuine network / auth errors so we don't silently lose progress.
    """
    try:
        raw = with_retry(
            api.hf_hub_download,
            repo_id=DST_REPO_ID,
            repo_type=REPO_TYPE,
            filename=LEDGER_REPO_PATH,
            token=HF_TOKEN,
            label="pull ledger",
            no_retry=(EntryNotFoundError,),   # 404 = ledger doesn't exist yet, don't retry
        )
        with open(raw) as f:
            data = json.load(f)
        # Normalise: deduplicate lists, ensure all keys exist
        data["uploaded"] = sorted(set(data.get("uploaded", [])))
        data["failed"]   = sorted(set(data.get("failed",   [])))
        data.setdefault("last_batch", -1)
        print(f"   Ledger pulled from Hub — "
              f"{len(data['uploaded'])} uploaded, {len(data['failed'])} failed.")
        return data
    except EntryNotFoundError:
        print("    No ledger on Hub yet — starting fresh.")
        return _empty_ledger()


def ledger_as_bytes(progress: dict) -> bytes:
    """Serialise ledger to UTF-8 bytes for inline commit upload."""
    # Always deduplicate before saving
    clean = {
        "uploaded":   sorted(set(progress["uploaded"])),
        "failed":     sorted(set(progress["failed"])),
        "last_batch": progress["last_batch"],
    }
    return json.dumps(clean, indent=2).encode("utf-8")


# ── helpers ────────────────────────────────────────────────────────────────────

def stem_from_repo_path(repo_path: str) -> str:
    return Path(repo_path).name.replace("_img.nii.gz", "")

def mask_repo_path(stem: str) -> str:
    return f"{MASKS_DIR}/{stem}_msk.nii.gz"
```

```python
# ── 5. Hub discovery helpers ───────────────────────────────────────────────────

def list_all_nifti_files() -> list:
    """Return all *.nii.gz repo-relative paths in IMAGES_DIR, sorted."""
    pattern = f"datasets/{SRC_REPO_ID}/{IMAGES_DIR}/*.nii.gz"
    all_files = with_retry(fs.glob, pattern, label="list source files")
    prefix    = f"datasets/{SRC_REPO_ID}/"
    rel_paths = sorted(f[len(prefix):] for f in all_files)
    print(f"  Found {len(rel_paths)} NIfTI file(s) in '{IMAGES_DIR}/'.")
    return rel_paths


def get_hub_mask_stems() -> set:
    """
    List masks already on the Hub.
    Used at startup to reconcile the ledger with reality.
    Returns empty set on failure (logged, not fatal).
    """
    try:
        pattern = f"datasets/{DST_REPO_ID}/{MASKS_DIR}/*.nii.gz"
        files   = with_retry(fs.glob, pattern, label="list hub masks")
        return {Path(f).name.replace("_msk.nii.gz", "") for f in files}
    except Exception as exc:
        print(f"    Could not list Hub masks (will rely on ledger only): {exc}")
        return set()
```

```python
# ── 6. Batch commit — masks + updated ledger in one shot ──────────────────────

def commit_batch(local_paths: list, repo_paths: list,
                 progress: dict, batch_idx: int) -> bool:
    """
    Push all masks for this batch AND the updated ledger as a SINGLE commit.

    Bundling the ledger into the same commit means that if the commit succeeds
    the Hub always has a ledger that exactly matches what was uploaded — there
    is no window where masks exist but the ledger hasn't been updated yet.

    Returns True on success, False after all retries exhausted.
    """
    # Build mask operations
    operations = [
        CommitOperationAdd(
            path_in_repo=remote,
            path_or_fileobj=str(local),
        )
        for local, remote in zip(local_paths, repo_paths)
    ]

    # Build ledger operation — inline bytes, no temp file needed
    operations.append(
        CommitOperationAdd(
            path_in_repo=LEDGER_REPO_PATH,
            path_or_fileobj=io.BytesIO(ledger_as_bytes(progress)),
        )
    )

    first = Path(repo_paths[0]).name
    last  = Path(repo_paths[-1]).name
    commit_msg = (
        f"Masks batch {batch_idx + 1} — "
        f"{len(local_paths)} file(s) [{first} … {last}]"
    )

    try:
        with_retry(
            api.create_commit,
            repo_id=DST_REPO_ID,
            repo_type=REPO_TYPE,
            commit_message=commit_msg,
            operations=operations,
            token=HF_TOKEN,
            label=f"commit batch {batch_idx + 1}",
        )
        print(f"  → 1 commit: {len(local_paths)} mask(s) + ledger.")
        return True
    except Exception as exc:
        print(f"     Commit failed after all retries: {exc}")
        print(f"     Local masks kept in {MASKS_OUT} — will retry on next run.")
        return False
```

```python
# ── 7. Segmentation ────────────────────────────────────────────────────────────

def _process_slice(sl: np.ndarray, W: int) -> np.ndarray:
    """
    Clean a single 2-D binary slice (H, W).

    If the two lungs are already separated, close and fill each blob
    independently using a conservative disk — no risk of a straight-line edge.

    If they are touching, find the natural intensity valley in the central band
    to locate the mediastinal gap, split there, process each half, recombine,
    then heal the cut edge with a tiny closing so no hard vertical line remains.
    """
    disk_sep     = morphology.disk(DISK_CLOSING_RADIUS)    # conservative — blobs already apart
    disk_touch   = morphology.disk(DISK_CLOSING_TOUCHING)  # stronger — needs to bridge fissures
    disk_heal    = morphology.disk(SEAM_HEAL_RADIUS)        # tiny — only smooths the seam

    labeled_sl = measure.label(sl)
    n_blobs    = labeled_sl.max()

    if n_blobs >= 2:
        # Already separated — process each blob independently, no split needed.
        result = np.zeros_like(sl)
        for lbl in range(1, n_blobs + 1):
            blob = labeled_sl == lbl
            if blob.sum() < MIN_BLOB_VOXELS:
                continue
            blob = morphology.binary_closing(blob, disk_sep)
            blob = ndi.binary_fill_holes(blob)
            result |= blob
        return result

    else:
        # Lungs are touching — find the natural valley in the central band.
        cx_lo    = W // 5
        cx_hi    = W - W // 5
        col_sums = sl[:, cx_lo:cx_hi].sum(axis=0)
        split    = cx_lo + int(np.argmin(col_sums))

        left  = sl.copy();  left[:, split:] = False
        right = sl.copy();  right[:, :split] = False

        for half in (left, right):
            half[:] = morphology.binary_closing(half, disk_touch)
            half[:] = ndi.binary_fill_holes(half)

        merged = left | right

        # Heal the hard vertical cut edge with a small closing on the full slice.
        merged = morphology.binary_closing(merged, disk_heal)
        merged = ndi.binary_fill_holes(merged)
        return merged


def segment_lungs(volume: np.ndarray) -> np.ndarray:
    """
    Generate a binary lung mask from an HU volume (H, W, Z) — shape preserved.

    Steps
    -----
    1. Threshold → air mask.
    2. Per-slice: clear_border → fill → smart L/R split (only if needed) →
       per-blob closing/fill → seam heal.
    3. 3-D connected components: keep largest N regions inside the lung Z band.
       Spatial filtering removes abdominal gas and table artifacts.
    4. Final per-slice fill + light 3-D closing → smooth surfaces.
    """
    H, W, n_slices = volume.shape

    # Step 1 – threshold
    mask = volume < HU_THRESHOLD

    # Step 2 – per-slice cleanup
    for i in range(n_slices):
        sl         = clear_border(mask[:, :, i])
        sl         = ndi.binary_fill_holes(sl)
        mask[:, :, i] = _process_slice(sl, W)

    # Step 3 – 3-D connected components with spatial filtering
    labeled = measure.label(mask)
    props   = measure.regionprops(labeled)

    if not props:
        warnings.warn("No connected components found — returning empty mask.")
        return mask.astype(bool)

    props.sort(key=lambda p: p.area, reverse=True)
    max_area = props[0].area
    Z        = n_slices

    # Anchor the lung Z band to the two largest candidate regions
    top_two     = [p for p in props if p.area >= max_area * MIN_LUNG_FRACTION][:2]
    z_centroids = [p.centroid[2] for p in top_two]
    z_center    = np.mean(z_centroids)
    z_band      = 0.40 * Z
    z_min_ok    = max(0,   z_center - z_band)
    z_max_ok    = min(Z-1, z_center + z_band)

    def is_lung_candidate(p):
        if p.area < max_area * MIN_LUNG_FRACTION:
            return False
        cz, cy, cx = p.centroid[2], p.centroid[0], p.centroid[1]
        if not (z_min_ok <= cz <= z_max_ok):   # rejects subdiaphragmatic gas
            return False
        margin = 0.05                           # rejects table-edge artifacts
        if cx < W * margin or cx > W * (1 - margin):
            return False
        if cy < H * margin or cy > H * (1 - margin):
            return False
        return True

    keep_labels = [p.label for p in props if is_lung_candidate(p)][:MAX_LUNG_REGIONS]

    if not keep_labels:
        warnings.warn("Spatial filtering removed all regions — falling back to top 2 by size.")
        keep_labels = [p.label for p in props[:2]]

    mask = np.isin(labeled, keep_labels)

    # Step 4 – final smoothing
    for i in range(n_slices):
        mask[:, :, i] = ndi.binary_fill_holes(mask[:, :, i])
    mask = morphology.binary_closing(mask, morphology.ball(CLOSING_RADIUS))
    for i in range(n_slices):
        mask[:, :, i] = ndi.binary_fill_holes(mask[:, :, i])

    return mask.astype(bool)
```

```python
# ── 8. Per-file processor ──────────────────────────────────────────────────────

def process_file(repo_path: str) -> Path:
    """
    Download → segment → save mask locally.
    The source NIfTI is deleted from the HF download cache immediately after
    the mask is written to prevent disk exhaustion across batches.
    Returns the local mask Path on success, raises on any failure.
    """
    stem     = stem_from_repo_path(repo_path)
    msk_path = MASKS_OUT / f"{stem}_msk.nii.gz"

    if msk_path.exists():
        print(f"  {stem}  →  SKIP (mask already on local scratch)")
        return msk_path

    # Download with retry
    local_src = with_retry(
        hf_hub_download,
        repo_id=SRC_REPO_ID,
        repo_type=REPO_TYPE,
        filename=repo_path,
        cache_dir=str(HF_CACHE),
        token=HF_TOKEN,
        label=f"download {stem}",
    )

    # Load — if this raises, the cached file may be corrupt; delete it so
    # the next run re-downloads rather than hitting the same corrupt copy.
    try:
        nii    = nib.load(local_src)
        volume = np.asarray(nii.dataobj, dtype=np.int16)    # already in HU, shape (H, W, Z)
        affine = nii.affine
    except Exception as exc:
        Path(local_src).unlink(missing_ok=True)
        raise RuntimeError(f"Failed to load NIfTI (corrupt cache? deleted): {exc}") from exc

    mask = segment_lungs(volume)

    nib.save(nib.Nifti1Image(mask.astype(np.uint8), affine), str(msk_path))
    print(f"  {stem}  →  OK  (shape {volume.shape}, mask voxels: {mask.sum():,})")

    # ── Disk hygiene: delete the source from HF cache immediately ──────────────
    # hf_hub_download stores files inside a nested blob dir; removing the whole
    # snapshot folder is the safest approach.
    try:
        src_path = Path(local_src)
        # Walk up to the snapshot root (…/<repo_hash>/snapshots/<commit>/) and remove it
        snapshot_root = src_path.parent
        if snapshot_root.name != "hf" and snapshot_root.exists():
            shutil.rmtree(snapshot_root, ignore_errors=True)
    except Exception:
        pass  # best-effort; disk will be cleaned by Colab anyway on session end

    # Free RAM explicitly before next file
    del volume, mask, nii
    gc.collect()

    return msk_path
```

```python
# ── 9. Main pipeline ───────────────────────────────────────────────────────────

def main():
    print("=" * 60)
    print(f"Source : {SRC_REPO_ID}")
    print(f"Dest   : {DST_REPO_ID}")
    print(f"Flow   : {IMAGES_DIR}/  →  {MASKS_DIR}/")
    print(f"Batch size: {BATCH_SIZE}")
    print("=" * 60)

    # ── A. Pull ledger from Hub (source of truth) ──────────────────────────────
    print("\n[Step 1] Pulling progress ledger from Hub…")
    progress = pull_ledger()
    already_uploaded: set = set(progress["uploaded"])

    # ── B. Discover source files ───────────────────────────────────────────────
    print("\n[Step 2] Discovering source files…")
    all_files = list_all_nifti_files()

    # ── C. Reconcile Hub mask folder against ledger ────────────────────────────
    # This catches masks that were committed before the ledger existed,
    # or uploaded by another process.
    print("\n[Step 3] Reconciling Hub mask folder with ledger…")
    hub_stems = get_hub_mask_stems()
    new_from_hub = hub_stems - already_uploaded
    if new_from_hub:
        print(f"  Found {len(new_from_hub)} mask(s) on Hub not in ledger — adding.")
        already_uploaded |= new_from_hub
        progress["uploaded"] = sorted(already_uploaded)
    else:
        print("  Ledger is consistent with Hub.")

    # ── D. Build todo list ─────────────────────────────────────────────────────
    todo = [fp for fp in all_files if stem_from_repo_path(fp) not in already_uploaded]
    print(f"\n  {len(todo)} file(s) to process ({len(already_uploaded)} already done).")

    if not todo:
        print("\n✅ All files already processed and uploaded. Nothing to do.")
        return

    # ── E. Process in batches ──────────────────────────────────────────────────
    batches = [todo[i:i + BATCH_SIZE] for i in range(0, len(todo), BATCH_SIZE)]
    print(f"  → {len(batches)} batch(es) of up to {BATCH_SIZE} file(s) each.\n")

    for batch_idx, batch in enumerate(batches):
        print(f"{'─' * 60}")
        print(f"BATCH {batch_idx + 1}/{len(batches)}  ({len(batch)} files)")
        print(f"{'─' * 60}")

        produced_local  = []
        produced_remote = []
        batch_failed    = []

        # Process each file
        for i, repo_path in enumerate(batch, 1):
            stem = stem_from_repo_path(repo_path)
            print(f"\n  [{i}/{len(batch)}] {Path(repo_path).name}")

            # If a mask is sitting on local scratch from a previous crashed run,
            # include it directly without reprocessing.
            local_msk = MASKS_OUT / f"{stem}_msk.nii.gz"
            if local_msk.exists():
                print(f"  {stem}  →  SKIP (found on local scratch from previous run)")
                produced_local.append(local_msk)
                produced_remote.append(mask_repo_path(stem))
                continue

            try:
                result = process_file(repo_path)
                produced_local.append(result)
                produced_remote.append(mask_repo_path(stem))
            except Exception as exc:
                print(f"   FAILED {stem}: {exc}")
                batch_failed.append(stem)
                # Add to failed set (deduplication happens in ledger_as_bytes)
                progress["failed"] = sorted(set(progress["failed"]) | {stem})

        # Report batch processing result
        if batch_failed:
            print(f"\n    {len(batch_failed)} file(s) failed this batch: {batch_failed}")

        if not produced_local:
            print(f"    No masks produced in batch {batch_idx + 1} — skipping commit.")
            # Still push updated ledger (has new failures recorded) as a standalone commit
            try:
                with_retry(
                    api.create_commit,
                    repo_id=DST_REPO_ID,
                    repo_type=REPO_TYPE,
                    commit_message=f"Update ledger — batch {batch_idx + 1} (all files failed)",
                    operations=[
                        CommitOperationAdd(
                            path_in_repo=LEDGER_REPO_PATH,
                            path_or_fileobj=io.BytesIO(ledger_as_bytes(progress)),
                        )
                    ],
                    token=HF_TOKEN,
                    label="ledger-only commit",
                )
            except Exception as exc:
                print(f"    Could not push ledger update: {exc}")
            continue

        # Update progress dict with newly produced stems before committing
        new_stems = {Path(p).name.replace("_msk.nii.gz", "") for p in produced_local}
        already_uploaded |= new_stems
        progress["uploaded"]   = sorted(already_uploaded)
        progress["last_batch"] = batch_idx

        # Commit masks + ledger to Hub in one atomic operation
        print(f"\n  ↑ Committing {len(produced_local)} mask(s) + ledger to Hub…")
        upload_ok = commit_batch(produced_local, produced_remote, progress, batch_idx)

        if upload_ok:
            # Only clean up local masks after confirmed commit
            for local_path in produced_local:
                try:
                    local_path.unlink()
                except Exception:
                    pass
            print(f"   Batch {batch_idx + 1} done — local scratch cleaned.")
        else:
            # Commit failed — roll back progress so this batch is retried
            already_uploaded -= new_stems
            progress["uploaded"]   = sorted(already_uploaded)
            progress["last_batch"] = batch_idx - 1
            print(f"    Batch {batch_idx + 1} commit failed. "
                  f"Local masks kept — will retry on next run.")

    # ── F. Final summary ───────────────────────────────────────────────────────
    print("\n" + "=" * 60)
    print("Pipeline complete.")
    print(f"  Uploaded (all runs) : {len(progress['uploaded'])}")
    print(f"  Failed   (all runs) : {len(set(progress['failed']))}")
    if progress["failed"]:
        print(f"  Failed stems        : {sorted(set(progress['failed']))}")
    print(f"  Ledger on Hub       : {LEDGER_REPO_PATH}")
    print("=" * 60)


main()
```
